```python
#prob- 1
import pandas as pd

# using Python List
marks_list = [78, 85, 92, 67, 88]

print("Python List:")
print(marks_list)

#  using Pandas Series
marks_series = pd.Series(marks_list)

print("\nPandas Series:")
print(marks_series)

# Advantages of Pandas over Python Lists
print("\nAdvantages of Pandas:")
print("1. Pandas provides powerful data analysis and manipulation functions.")
print("2. Pandas makes filtering, sorting, and statistical calculations easier.")
print("3. Pandas can easily read and write CSV, Excel, JSON, and other data files.")
```

```python
#prob-14
import pandas as pd

sales_data = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Tablet", "Keyboard", "Mouse", "Monitor"],
    "Sales": [50000, 30000, 25000, 10000, 5000, 20000],
    "Quantity": [5, 10, 8, 20, 30, 12]
})

sales_data.to_excel("sales.xlsx", index=False)

sales = pd.read_excel("sales.xlsx")

print(sales.tail(5))
```
