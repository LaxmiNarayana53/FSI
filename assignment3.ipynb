```python
# Ask the user to enter today's memory
memory = input("Enter today's memory: ")

# Append the memory to diary.txt
with open("diary.txt", "a") as file:
    file.write(memory + "\n")

# Read and display all memories
print("\nSaved Memories:")
with open("diary.txt", "r") as file:
    print(file.read())
```

```python
# Create orders.txt and write sample data
file = open("orders.txt", "w")
file.write("Laptop\n")
file.write("Mouse\n")
file.write("Keyboard\n")
file.write("Headphones\n")
file.close()

# Read the file
count = 0
file = open("orders.txt", "r")

print("Items Purchased:")
for item in file:
    print(item.strip())
    count += 1

file.close()

print("Total Orders:", count)
```

```python
password = input("Enter your password: ")

upper = 0
lower = 0
digit = 0

for ch in password:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
    elif ch.isdigit():
        digit += 1

print("Uppercase letters:", upper)
print("Lowercase letters:", lower)
print("Digits:", digit)

if len(password) >= 8 and upper > 0 and lower > 0 and digit > 0:
    print("Password Strength: Strong")
else:
    print("Password Strength: Weak")
```

```python
message = "Python is Fun and Python is Powerful"

# Count occurrences of "Python"
count = message.count("Python")

# Convert to uppercase
upper_msg = message.upper()

# Replace "Powerful" with "Amazing"
new_msg = message.replace("Powerful", "Amazing")

# Display results
print("Number of times 'Python' appears:", count)
print("Uppercase Message:", upper_msg)
print("Modified Message:", new_msg)
```

```python
def withdraw(balance, amount):
    if amount <= balance:
        balance = balance - amount
        return balance
    else:
        return "Insufficient Balance"

balance = float(input("Enter Account Balance: "))
amount = float(input("Enter Withdrawal Amount: "))

result = withdraw(balance, amount)

if isinstance(result, str):
    print(result)
else:
    print("Remaining Balance:", result)
```

```python
def calculate_grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 75:
        return "B"
    elif mark >= 60:
        return "C"
    else:
        return "F"

mark = int(input("Enter student marks: "))
grade = calculate_grade(mark)

print("Grade:", grade)
```

```python
def calculate_bill(price, quantity):
    total = price * quantity
    return total

price = float(input("Enter price of item: "))
quantity = int(input("Enter quantity: "))

bill = calculate_bill(price, quantity)

print("\n----- BILL -----")
print("Price:", price)
print("Quantity:", quantity)
print("Total Bill: ₹", bill)
```

```python
# Lambda function to calculate 10% discount
discount = lambda price: price - (price * 10 / 100)

price = float(input("Enter product price: "))

print("Price after 10% discount:", discount(price))
```

```python
# Lambda function to calculate rectangle area
area = lambda length, width: length * width

length = float(input("Enter length: "))
width = float(input("Enter width: "))

print("Area of Rectangle:", area(length, width))
```

```python
# Lambda function to check speed
speed_check = lambda speed: "Overspeed" if speed > 80 else "Normal Speed"

speed = int(input("Enter speed: "))

print(speed_check(speed))
```
