```python
seats = {
    "A1": "Available",
    "A2": "Booked",
    "A3": "Available",
    "A4": "Available"
}

seat = input("Enter Seat Number: ")

if seat in seats:
    if seats[seat] == "Available":
        seats[seat] = "Booked"
        print("Seat Booked Successfully")
    else:
        print("Seat Already Booked")
else:
    print("Invalid Seat Number")

print("\nUpdated Seat Status:")
for key, value in seats.items():
    print(key, ":", value)
```

```python
items = {
    "Milk": 50,
    "Bread": 40,
    "Eggs": 70,
    "Rice": 500
}

total = 0

while True:
    item = input("Enter item name (or 'done' to finish): ")

    if item.lower() == "done":
        break

    if item in items:
        total += items[item]
        print(item, "added to bill.")
    else:
        print("Item not found.")

print("\nFinal Bill = ₹", total)
```

```python
# Create a student dictionary
student = {
    "Name": "Rahul",
    "Age": 20,
    "Course": "AI & ML",
    "City": "Hyderabad"
}

# Display student details
print("Student Details:")
for key, value in student.items():
    print(key, ":", value)

# Display dictionary methods
print("\nKeys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())

# Update city
student.update({"City": "Warangal"})

# Display updated dictionary
print("\nUpdated Student Details:")
for key, value in student.items():
    print(key, ":", value)
```

```python
systems = {
    "Engine": "OK",
    "Fuel": "OK",
    "Navigation": "FAIL",
    "Communication": "OK"
}

with open("mission_log.txt", "w") as file:

    for system, status in systems.items():

        if status == "FAIL":
            print("Failed System:", system)
            file.write(system + " : FAIL\n")
            print("Mission Aborted")
            break

    else:
        file.write("Launch Successful")
        print("Launch Successful")
```
