```python
#problem - 1
scores = [520, 780, 430] 
for i in range(len(scores)):
    for j in range(len(scores)-1):
        if scores[j] < scores[j+1]:
           temp=scores[j]
           scores[j]=scores[j+1]
           scores[j+1]=temp
print('first rank',scores[0])
print('second rank',scores[1])
print('third rank',scores[2])
p1,p2,p3=scores[0],scores[1],scores[2]
if (p1==p2 or p2==p3 or p3==p1):
    print('tie detected')
```
